```python
def print_msg(number):
    def printer():

      nonlocal number
      number = 3
      print(number)

    printer()
    print(number)
print_msg(9)
```

Decorator

```python
from functools import wraps
def my_Decorator(func):
  @wraps(func)
  def wrapper():
    print('before the func')
    func()
    print('after the func')
  return wrapper
@my_Decorator
def printer_name():
  print('David')

printer_name()
print("__name", printer_name.__name__)
print("__doc", printer_name.__doc__)
```

```python
import functools
import time

def timer(func):
    @functools.wraps(func)
    def wrapper_timer(*args, **kwargs):
      start_time = time.perf_counter()
      value = func(*args, **kwargs)
      end_time = time.perf_counter()
      run_time = end_time - start_time
      print(f"finish {func.__name__} in {run_time} secs")
      return value
    return wrapper_timer
@timer
def waste_some_time(num_time):
    for _ in range(num_time):
      total = sum(i**2 for i in range(100000))
    return total
waste_some_time(44)
```

**@Property**

```python
class  Person:
    def __init__(self, brith_year):
        self._brith_year = brith_year

    def get_age(self):
        from datetime import date
        return date.today().year - self._brith_year

    age = property(get_age)
p = Person(2000)
p.age
```

```python
from os import set_inheritable
class Rectangle:
    def __init__(self, w, h):
        self._w = w
        self._h = h

    @property
    def width(self):
      return self._w
    @width.setter
    def width(self, value):
        if value <= 0:
          raise ValueError("width must be > 0")
        self._w = float(value)

    @width.deleter
    def width(self):
      raise AttributeError('width can not be delete')

    @property
    def area(self):
        return self._w * self._h

r = Rectangle(5, 4)
print(r.area)
print(r.width)
r.width = 40
print(r.width)
print(r.area)

```

```python
class Account:
    def __init__(self, balance): self._balance = balance
    @property
    def balance(self): return self._balance
    @balance.setter
    def balance(self, v):
      if v < 0: raise ValueError('negative')
      self._balance = v
class LockedAccount(Account):
    @property
    def balance(self):
       return super().balance
    @balance.setter
    def balance(self, v):
      raise PermissionError('Account is Locked')

a1 = Account(2900)
print(a1.balance)
a1.balance = 3000
print(a1.balance)

a2 = LockedAccount(a1)
print(a2.balance)
# a2.balance = 2000

```

**@classmethod and @staticmethod**

```python
from datetime import date

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    @classmethod
    def fromBrithYear(cls, name, year):
        return cls(name, date.today().year - year)

    @staticmethod
    def isAdult(age):
        return age > 18
person1 = Person('Bob', 21)
person2 = Person.fromBrithYear('Tom', 1998)

print(person1.age)
print(person2.age)

print(Person.isAdult(22))
```

**inner class**

```python
class Student:

  def __init__(self, name, enrollno):
      self.name = name
      self.enrollno = enrollno
      self.lap = self.Laptop()

  def show(self):
      print(f'my name is {self.name}, and my enrollno is {self.enrollno}')
      self.lap.show()

  class Laptop:

      def __init__(self):
          self.brand = "Apple"
          self.cpu = "M2"
          self.ram = 16
      def show(self):
          print(self.brand, self.cpu, self.ram)
s1 = Student('Toms', 5435256)
s1.show()
l1 = Student.Laptop()
l1.brand = "HP"
l1.cpu = "i9"
s1.lap = l1
s1.show()
```
