```python
mask = df["math"] < 75
df.loc[mask, "math"] = (df.loc[mask, "math"] + 5).clip(upper=100)

df.loc[df["class"]=="A2", ["english"]] += 2
df["total"] = df["math"] + df["english"] + df["cs"]
df
```

```python
df["math_z"] = (df["math"] - df["math"].mean()) / df["math"].std()

df["class"] = df["class"].replace({
    "A1": "Class-A",
    "A2": "Class-B"
})

bonus = {"ann": 2, "ben":1, "cici":3}
df["bonus"] = df.index.map(bonus).fillna(99).astype(int)
df
```

```python
df = df.rename(columns={"english": "eng"})

df["eng"] = df["eng"].astype("int64")

df.loc["david", "eng"] = None
df["eng"] = df["eng"].fillna(df["eng"].mean())
df
```

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["ann", "ben", "cici", "dave", "eric", "fiona", "gina"],
    "class": ["A1", "A1", "A2", "A2", "A2", "A1", "A3"],
    "math": [88, 83, 90, 76, 78, 91, 85],
    "eng": [95, 86, 90, 80, 85, 92, 84],
    "cs": [96, 82, 94, 79, 84, 93, 87]
}).set_index("name")

df["avg"] = df[["math", "eng", "cs"]].mean(axis=1).round(1)
df
```

```python
print(df.describe())
print(df['math'].mean())
print(df['math'].median())
print(df['math'].mode())
print(df['math'].std())
print(df['math'].min(),"-----",df['math'].max())
print(df['avg'].sum())
```

```python
df_sorted = df.sort_values(by="avg", ascending=False)
print(df_sorted)

df_multi = df.sort_values(by=['class', 'avg'], ascending=[True, False])
print(df_multi)
```

```python
group_mean = df.groupby('class')['avg'].mean()
print(group_mean)
group_stats = df.groupby('class')[['math', 'cs', 'eng']].mean().round(1)
print(group_stats)
```

```python
print(df.groupby('class')['avg'].agg(['mean','min','max','count']))

agg_multi = df.groupby('class').agg({
    "math": ["mean", "max"],
    "eng": ["mean", "min"],
    "cs": "std"
})
print(agg_multi)
```

```python
def score_range(x):
    return x.max() - x.min()
print(df.groupby('class')[['math','eng','cs']].agg(score_range))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline

df = pd.DataFrame({
    "name": ["ann","ben","cici","dave","eric","fiona","gina"],
    "class": ["A1","A1","A2","A2","A2","A1","A3"],
    "math": [88,83,90,76,78,91,85],
    "eng":  [95,86,90,80,85,92,84],
    "cs":   [96,82,94,79,84,93,87]
}).set_index("name")

df["avg"] = df[["math","eng","cs"]].mean(axis=1)
df
```
